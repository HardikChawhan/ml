# 22 - Single Perceptron for AND / OR

Train a single-layer perceptron to learn the AND and OR logic functions using a small CSV sample.

**Problem Statement**

Implement a single perceptron that learns the `AND` and `OR` boolean functions from a CSV containing two binary inputs.

**Understanding of Dataset**

We use a tiny CSV with columns `x1`, `x2`, and `label`. For `AND`, label is x1 & x2; for `OR`, label is x1 | x2. The notebook generates and saves `perceptron_logic_sample.csv` and demonstrates training for both gates.

**Theory**

A perceptron computes y = step(w0 + w1*x1 + w2*x2). We update weights by the perceptron learning rule: w <- w + lr*(target - pred)*x (including bias as x0=1). Single-layer perceptron can learn any linearly separable function (AND and OR are separable).

**Mathematical Representation**

Prediction: y = 1 if w^T x + b > 0 else 0.
Update: w <- w + lr*(t - y)*x; b <- b + lr*(t - y)

**Pseudocode**

1. Load CSV -> Clean -> (no scaling needed for binary inputs)
2. Initialize weights and bias
3. For epochs: for each sample, compute prediction, update weights
4. Evaluate accuracy and print weights

```python
# Simple perceptron implementation for AND and OR with comments
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score

# Create and save a small CSV for logic gates (x1, x2)
df = pd.DataFrame({
    'x1': [0,0,1,1],
    'x2': [0,1,0,1],
})
# We'll store two labels so you can switch: 'AND' and 'OR'
df['AND'] = (df['x1'] & df['x2']).astype(int)
df['OR'] = (df['x1'] | df['x2']).astype(int)
df.to_csv('perceptron_logic_sample.csv', index=False)
print('Saved perceptron_logic_sample.csv with columns x1,x2,AND,OR')
df
```

```python
# === Load CSV and quick cleaning ===
data = pd.read_csv('perceptron_logic_sample.csv')
# If your exam CSV has different column names, rename to ['x1','x2'] and choose label column.
print('Data:')
print(data)

# We'll define a simple perceptron trainer function
def train_perceptron(X, y, lr=0.1, epochs=10):
    # X: (n_samples, n_features) without bias column. We'll add bias internally.
    n_samples, n_features = X.shape
    # initialize weights (including bias) to zeros for simplicity
    w = np.zeros(n_features)
    b = 0.0
    for epoch in range(epochs):
        for i in range(n_samples):
            xi = X[i]
            linear = np.dot(w, xi) + b
            y_pred = 1 if linear > 0 else 0
            update = lr * (y[i] - y_pred)
            # update weights and bias
            w += update * xi
            b += update
    return w, b

def predict_perceptron(X, w, b):
    linear = np.dot(X, w) + b
    return (linear > 0).astype(int)

# Prepare input matrix
X = data[['x1','x2']].values.astype(float)
# No scaling required for binary inputs, but leave placeholder if needed.

# Train for AND
y_and = data['AND'].values
w_and, b_and = train_perceptron(X, y_and, lr=0.1, epochs=20)
pred_and = predict_perceptron(X, w_and, b_and)
acc_and = accuracy_score(y_and, pred_and)
print('AND weights:', w_and, 'bias:', b_and, 'accuracy:', acc_and)
print('Predictions (AND):', pred_and, 'Targets:', y_and)

# Train for OR
y_or = data['OR'].values
w_or, b_or = train_perceptron(X, y_or, lr=0.1, epochs=20)
pred_or = predict_perceptron(X, w_or, b_or)
acc_or = accuracy_score(y_or, pred_or)
print('OR weights:', w_or, 'bias:', b_or, 'accuracy:', acc_or)
print('Predictions (OR):', pred_or, 'Targets:', y_or)
```
